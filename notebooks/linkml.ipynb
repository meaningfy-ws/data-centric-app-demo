# Why developer-friendly?

- Rich modeling languages have steep learning curves🧗‍♂️
- People are more comfortable with familiar syntax💡
- Development tooling for RDF & OWL is lacking/subpar😓
- Single Source of Truth (SSoT) enables model-driven development🚀

# YAML as a Linked Modeling Language

- Simple, straightforward indentation-based list syntax✅
- Used often for application/project configuration⚙️
- Mature tooling ecosystem for many programming languages🧰

```yaml
classes:
  Person:
    class_uri: schema:Person
    attributes:
      id:
        identifier: true
      name:
        slot_uri: schema:name
        required: true
```

# Multiple target formats

Generators for many modeling and programming artefacts

- Schema frameworks (JSON Schema, ProtoBuf, GraphQL)
- Linked data standards (RDF, OWL, SHACL, ShEx, SPARQL, JSON-LD)
- Documentation & diagramming (Markdown/MKDocs, Mermaid, PlantUML, YUML)
- Programming languages (Python/Pydantic, Java, TypeScript)

# Handy knowledge engineering tools

- Generate OWL with `gen-owl $MODEL_FILE > $OUTPUT_OWL`

- Generate SHACL with `gen-shacl $MODEL_FILE > $SHACL_FILE`

- Convert data with `linkml-convert -s $MODEL_FILE $YAML_DATA -o $TURTLE_DATA`

- Validate data with `linkml-validate -s $MODEL_FILE $DATA_FILE`
